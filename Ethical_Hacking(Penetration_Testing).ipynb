# 5 STAGES OF PENETRATION TESTING (Do not hack what you have not been authorized to hack.)
1. INFORMATION GATHERING/RECONNAISANCE: This is the act of discovering information information about your target. This can be ACTIVE which entails interacting with the target or PASSIVE which involves not interacting with the target. For instance, visitng xyz.com website is you actively gathering information about it as a target. However, getting information about xyz.com from other websites or documents that talk about xyz.com means you did not interact directly with it and hence passive.

2. SCANNING: This can get you in trouble if you do it without permission. Scanning is a deeper form of information gathering using technical tools to find openings in the target and in the system that you're attacking. These opening can be gateways, open ports, operating system that the target runs etc.

3. GAINING ACCESS/EXPLOITATION: This is the step where we actually hack the target. We use information gathered in stages one and two to take control of any number of devices. Gaining access to target devices allows us to steal data from their system or to use those devices to attack other devices on the same network. Usually, after this step, you canc consider penetration tests to be successful since you managed to gain access to the tatget system. However , this is not he last step of a penetration test.

4. MAINTAINING ACCESS/PERSISTENCE: This step is usually optional. This is commonly done by installing backdoors and planting root kits. Backdoor and root kits are simple programs that will allow us to gain access to that target whenever we want without the need to exploit it again. Just connect the backdoor to the system and voila.

5. COVERING TRACKS: Covering tracks is simply removing all evidence that an attack took place. This can involve deleting or editing logs, deleting or hiding files or basically reverting any changes to the system while the attack took place.

# INFORMATION GATHERING
This is the act of gathering data on our target. It might be any type of data that we might find useful for the future attack.
- Active Information Gathering/Reconnaisance: Here we will use our kali machine and try to get as much information about our target as we can. It could be a target website or a network or perhaps and entire company. We will be interacting directly with our target. This could be exchanging packets with the target by visiting and enumerating their website or talking to an emaployee that works there. We could call employees at the target to gather certain key information. This could be legal to an extent. If you perform some advanced scans or OS fingerprinting on the target, you most likely wont get in trouble but you shouldn't still do it especially not without permission. Active information gathering provides much more important data than passive information gathering since we are directly interacting with the target.n
- Passive Information Gathering/Reconnaisance: This means that any information we are gathering about a target will go through the middle source which can be a website, search engine or even a person. For instance, if we want to find out something about a target we could google about this target and it is considered passive information gathering because we are not intereacting directly with the source.

# Whatweb Stealthy Scan
- Here we will discuss Whatweb. The tool is used to gather information and scan any website on the internet. It is primarily used to scan websites, since this tool recognizes web technologies including web servers, embedded devices, javascript libraries etc. The default level of this tool called 'stealthy' is the fastest and requires only one HTTP request of a website. This is suitable for scanning public websites. The other modes are more aggressive and should only be performed during penetration tests. You cannot use the more aggressive scans on websites you are not permitted to scan.
